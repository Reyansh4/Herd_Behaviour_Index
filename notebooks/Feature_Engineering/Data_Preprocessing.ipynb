```python
import os
import pandas as pd
from pathlib import Path
```

```python
data_path = Path("../../data/processed/herd_mentality_events_processed_final.csv").resolve()
processed_path = Path("../../data/gold_layer/herd_mentality_events_gold_data_v1.csv").resolve()
```

```python
df = pd.read_csv(data_path)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df = df.rename(columns={"M": "Magnitude(M)", "S": "Spread(S)", "I": "Intensity(I)", "D": "Duration(D)", "R": "Outcome/Impact(R)"})
df
```

```python
magnitude_value_map = {
    "1.0": "Minimal",
    "2.0": "Very Low",
    "3.0": "Low",
    "4.0": "Moderate-Low",
    "5.0": "Moderate",
    "6.0": "Moderate-High",
    "7.0": "High",
    "8.0": "Very High",
    "9.0": "Extreme",
    "10.0": "Historic"
}

spread_value_map = {
    "1.0": "Local",
    "2.0": "Multi-Local",
    "3.0": "City-Level",
    "4.0": "Regional",
    "5.0": "Multi-Regional",
    "6.0": "National",
    "7.0": "Bi-National",
    "8.0": "Multi-National",
    "9.0": "Continental",
    "10.0": "Global"
}

intensity_value_map = {
    "1.0": "Minimal",
    "2.0": "Very Low",
    "3.0": "Low",
    "4.0": "Moderate-Low",
    "5.0": "Moderate",
    "6.0": "Moderate-High",
    "7.0": "High",
    "8.0": "Very High",
    "9.0": "Critical",
    "10.0": "Maximum"
}

duration_value_map = {
    "1.0": "Momentary",
    "2.0": "Very Short",
    "3.0": "Short",
    "4.0": "Moderate-Short",
    "5.0": "Moderate",
    "6.0": "Long",
    "7.0": "Extended",
    "8.0": "Very Long",
    "9.0": "Multi-Year",
    "10.0": "Prolonged"
}

outcome_value_map = {
    "1.0": "Negligible",
    "2.0": "Very Low",
    "3.0": "Low",
    "4.0": "Moderate-Low",
    "5.0": "Moderate",
    "6.0": "Strong",
    "7.0": "Significant",
    "8.0": "Major",
    "9.0": "Transformational",
    "10.0": "Historic"
}

# Convert float64 values to strings before mapping (to match dictionary string keys)
df['Magnitude_label'] = df['Magnitude(M)'].astype(str).map(magnitude_value_map)
df['Spread_label'] = df['Spread(S)'].astype(str).map(spread_value_map)
df['Intensity_label'] = df['Intensity(I)'].astype(str).map(intensity_value_map)
df['Duration_label'] = df['Duration(D)'].astype(str).map(duration_value_map)
df['Outcome_label'] = df['Outcome/Impact(R)'].astype(str).map(outcome_value_map)
df
```

```python
df.to_csv(processed_path, index=False)
```
